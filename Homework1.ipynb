Площадь треугольника по формуле Герона

```python
import math

sideA = 0.3
sideB = 0.4
sideC = 0.5
p = (sideA + sideB + sideC)/2
area_triangle = math.sqrt(p * (p - sideA) * (p - sideB) * (p - sideC))

print("Area of a triangle:", area_triangle)
```

Написать программу, которая вычислит и выведет на экран сумму за покупку

```python
price_apple = 2
number_apples = 6
sum_apples = price_apple * number_apples

print ("The price for six apples:", sum_apples, "usd")
```

Вычислите и выведите на экран сумму оплаты за топливо для поездки

```python
price_liter_fuel = 1.2
way_100 = 8
distance = 120

fuel_liters = (way_100 / 100) * distance
fuel_cost = fuel_liters * price_liter_fuel

print("Total fuel cost:", fuel_cost, "usd")
```
