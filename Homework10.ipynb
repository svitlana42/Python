```python
def users(purchases):
  result = {}

  for name, product in purchases:
    if name not in result:
      result[name] = []

    result[name].append(product)

  return result

purchases =  [("Alice", "Book"), ("Alice", "Pen"), ("Bob", "Book")]

print(users(purchases))


```

```python
import statistics

def emissions(numbers):

  average = sum(numbers) / len(numbers)
  deviation = statistics.stdev(numbers)

  emissions_1 = []

  for number in numbers:
    if abs(number - average) > 3 * deviation:
      emissions_1.append(number)

  return emissions_1

numbers = [3600, 2578, 3152, 6300, 2195, 14800, 3378, 2003, 2948, 2080, 2040, 3232]

print(emissions(numbers))
```

```python
def quarter(sales):
  q1 = sum(sales[0:3])
  q2 = sum(sales[3:6])
  q3 = sum(sales[6:9])
  q4 = sum(sales[9:12])

  quarters = [q1, q2, q3, q4]

  max_sellers = max(quarters)

  if max_sellers == q1:
    return "Q1"
  elif max_sellers == q2:
    return "Q2"
  elif max_sellers == q3:
    return "Q3"
  else:
    return "Q4"

sales = [1200, 1300, 1500, 2000, 2100, 2500, 3000, 3100, 2800, 2200, 2300, 2100]

print(quarter(sales))

```

```python
def all_users(users):

  result = []

  for user in users:
    if user["login"] != "" and user["country"] != "" and user["email"] != "":
      result.append(user)

  return result

users = [
    {"login": "Alice", "country": "USA","email": "alice@mail.com"},
    {"login": "Bob", "country": "","email": "bobby@mail.com"},
    {"login": "Charlie", "country": "USA","email": ""}
]

print(all_users(users))
```
