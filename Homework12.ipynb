```python
class Product:

    def __init__(self, name, price, description, size):
        self.name = name
        self.price = price
        self.description = description
        self.size = size

    def __str__(self):
        return f"{self.name} - {self.price} €, {self.description}, size: {self.size}"

product1 = Product ("Laptop", 1500, "gaming laptop", "35 * 25 * 2 cm")

product2 = Product ("Mouse" , 200, "wireless mouse", "12 * 6 * 4 cm")

print(product1)
print(product2)

```

```python
class Customer:

    def __init__(self, surname, name, patronymic, phone, email):
        self.surname = surname
        self.name = name
        self.patronymic = patronymic
        self.phone = phone
        self.email = email

    def __str__(self):
        return f"{self.surname} {self.name} {self.patronymic}, phone: {self.phone}, e-mail: {self.email}"

customer = Customer("Ilchenko", "Svitlana", "Nikolaevna", "+380668859356", "ilchenkos65@gmail.com")

print(customer)
```

```python
class Order:

    def __init__(self, customer, products):
        self.customer = customer
        self.products = products
    def price(self):
        total = 0

        for product in self.products:
            total += product.price

        return total

    def __str__(self):
        result = f"Customer: {self.customer}\n"
        result += "Products:\n"

        for product in self.products:
            result += f"- {product.name}: {product.price} €\n"

        result += f"Total cost: {self.price()} €"

        return result

order = Order(customer, [product1, product2])

print(order)
```
