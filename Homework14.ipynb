```python
import pandas as pd

cars = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/cars_data_set.csv")

print(cars.head())
print(cars.info())
```

2

```python
cars = cars.rename(columns={
    "Make": "make",
    "Model": "model",
    "Year": "year",
    "Engine Fuel Type": "fuel_type",
    "Engine HP": "horsepower",
    "Engine Cylinders": "cylinders",
    "Transmission Type": "transmission",
    "Driven_Wheels": "drive_wheels",
    "Number of Doors": "doors",
    "Market Category": "category",
    "Vehicle Size": "vehicle_size",
    "Vehicle Style": "body_type",
    "highway MPG": "highway_mpg",
    "city mpg": "city_mpg",
    "Popularity": "popularity",
    "MSRP": "price"
})

print(cars.head())
```

3

```python
print(cars.isnull().sum())
```

```python
cars.loc[cars["fuel_type"].isin(["electric"]), "cylinders"] = 0

cars["horsepower"] = cars["horsepower"].fillna(cars["horsepower"].median())

cars["cylinders"] = cars["cylinders"].fillna(cars["cylinders"].median())

cars["doors"] = cars["doors"].fillna(cars["doors"].median())

cars["category"] = cars["category"].fillna("Unknown")

print(cars.isnull().sum())
```

4

```python
cars["year"] = cars["year"].astype(int)
cars["horsepower"] = cars["horsepower"].astype(int)
cars["cylinders"] = cars["cylinders"].astype(int)
cars["doors"] = cars["doors"].astype(int)
cars["highway_mpg"] = cars["highway_mpg"].astype(int)
cars["city_mpg"] = cars["city_mpg"].astype(int)
cars["popularity"] = cars["popularity"].astype(int)
cars["price"] = cars["price"].astype(int)

cars["make"] = cars["make"].astype("category")
cars["model"] = cars["model"].astype("category")
cars["fuel_type"] = cars["fuel_type"].astype("category")
cars["transmission"] = cars["transmission"].astype("category")
cars["drive_wheels"] = cars["drive_wheels"].astype("category")
cars["category"] = cars["category"].astype("category")
cars["vehicle_size"] = cars["vehicle_size"].astype("category")
cars["body_type"] = cars["body_type"].astype("category")

print(cars.info())
```

5

```python
cars["price_per_hp"] = cars["price"] / cars["horsepower"]

print(cars[["make", "model", "price", "horsepower", "price_per_hp"]].head())
```

6

```python
bmw = cars[cars["make"] == "BMW"]

bmw = bmw.sort_values(by="price", ascending=False)

print(bmw[["make", "model", "year", "price"]].head())
```
