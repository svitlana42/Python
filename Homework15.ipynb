```python
import pandas as pd

apartments = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/appartments_data_set.csv")

print(apartments.head())
print(apartments.info())
print(apartments.shape)
```

2

```python
apartments = apartments.drop(columns=["Unnamed: 0", "is_apartment", "days_exposition"])

apartments["balcony"] = apartments["balcony"].fillna(0)

apartments["ceiling_height"] = apartments["ceiling_height"].fillna(apartments["ceiling_height"].median())

apartments["floors_total"] = apartments["floors_total"].fillna(apartments["floors_total"].median())

apartments["living_area"] = apartments["living_area"].fillna(apartments["living_area"].median())

apartments["kitchen_area"] = apartments["kitchen_area"].fillna(apartments["kitchen_area"].median())

print(apartments.info())

apartments["price_per_sqm"] = apartments["last_price"] / apartments["total_area"]

print(apartments.head())
```

3

```python
print("Average price per square meter:", apartments["price_per_sqm"].mean())

print("Median price per square meter:", apartments["price_per_sqm"].median())

print("Standard declension:", apartments["price_per_sqm"].std())

print("Minimum price:", apartments["price_per_sqm"].min())

print("Maximum price:", apartments["price_per_sqm"].max())

Q1 = apartments["price_per_sqm"].quantile(0.25)
Q3 = apartments["price_per_sqm"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

4

```python
kyiv = apartments[apartments["locality_name"] == "Kyiv"]

print("Number of rooms in the apartment most sold in Kyiv:", kyiv["rooms"].mode()[0])
```

5

```python
irpin = apartments[(apartments["locality_name"] == "Irpin") & (apartments["rooms"] == 1)]

print("Average price of a one-room apartment in Irpin:", irpin["last_price"].mean())
```

```python
print("Number of studio apartments:", apartments["studio"].sum())

studio_percent = apartments["studio"].sum() / apartments.shape[0] * 100

print("Percentage of studio apartments:", studio_percent)
```
