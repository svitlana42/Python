```python
import pandas as pd

apartments = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/appartments_data_set.csv")

apartments.head()
```

2

```python
cities = apartments["locality_name"].unique()

max_count = 0
popular_city = ""

for city in cities:
    count = len(apartments[apartments["locality_name"] == city])

    if count > max_count:
        max_count = count
        popular_city = city

print("The city where the most apartments are sold:", popular_city)
print("Number of apartments:", max_count)
```

3

```python
studios = apartments[apartments["studio"] == True]

cities = studios["locality_name"].unique()

max_count = 0
popular_city = ""

for city in cities:
    count = len(studios[studios["locality_name"] == city])

    if count > max_count:
        max_count = count
        popular_city = city

print("The city where studio apartments are most popular:", popular_city)
print("Number of studio apartments:", max_count)
```

```python
cities = apartments["locality_name"].unique()
rooms = sorted(apartments["rooms"].unique())

table = []

for city in cities:
    city_apartments = apartments[apartments["locality_name"] == city]

    city_row = []

    for room in rooms:
        count = len(city_apartments[city_apartments["rooms"] == room])
        city_row.append(count)

    table.append(city_row)

rooms_table = pd.DataFrame(table, index=cities, columns=rooms)

rooms_table
```

В данных есть квартиры с необычным количеством комнат: 0 и больше 10. Такие объявления встречаются редко, но удалять их без дополнительной проверки не стоит

```python
apartments["price_per_sqm"] = apartments["last_price"] / apartments["total_area"]

kyiv = apartments[apartments["locality_name"] == "Kyiv"]

rooms = sorted(kyiv["rooms"].unique())

table = []

for room in rooms:
    room_apartments = kyiv[kyiv["rooms"] == room]

    average_price = room_apartments["price_per_sqm"].mean()

    table.append([room, average_price])

rooms_table = pd.DataFrame(table, columns=["Number of rooms", "Average price per м²"])

rooms_table
```

```python
kyiv[kyiv["rooms"] == 0].head(10)
```

В данных значение 0 обозначает квартиры без отдельных комнат. Большинство таких квартир являются студиями.

```python
heights = sorted(kyiv["ceiling_height"].dropna().unique())

table = []

for height in heights:
    height_apartments = kyiv[kyiv["ceiling_height"] == height]

    average_price = height_apartments["price_per_sqm"].mean()

    table.append([height, average_price])

heights_table = pd.DataFrame(table, columns=["Ceiling height", "Average price per м²"])

heights_table
```

```python
unusual_heights = kyiv[
    (kyiv["ceiling_height"] < 2) |
    (kyiv["ceiling_height"] > 5)
]

unusual_heights[["ceiling_height", "total_area", "rooms", "price_per_sqm"]]
```

В данных есть квартиры с необычно низкой и высокой высотой потолков. Возможно, при заполнении некоторых объявлений были допущены ошибки. Поэтому такие значения нужно учитывать с осторожностью.
