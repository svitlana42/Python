```python
import pandas as pd

apartments = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/appartments_data_set.csv")

apartments.head()
```

2

```python
apartments_2023 = apartments[
    (apartments["first_day_exposition"] >= "2023-01-01") &
    (apartments["first_day_exposition"] <= "2023-12-31")
]

two_rooms = apartments_2023[apartments_2023["rooms"] == 2]

cities = two_rooms["locality_name"].unique()

max_count = 0
popular_city = ""

for city in cities:
    count = len(two_rooms[two_rooms["locality_name"] == city])

    if count > max_count:
        max_count = count
        popular_city = city

print("Two-bedroom flats accounted for the largest share of sales in 2023:", popular_city)
print("Number of flats:", max_count)
```

В 2023 году больше всего объявлений о продаже двухкомнатных квартир было опубликовано в городе Киев — 1322 объявлений.
В датасете отсутствует информация о фактических продажах, поэтому анализ основан на датах публикации объявлений.

3

```python
apartments["year"] = apartments["first_day_exposition"].str[:4]

kyiv_three_rooms = apartments[
    (apartments["locality_name"] == "Kyiv") &
    (apartments["rooms"] == 3)
]

years = sorted(kyiv_three_rooms["year"].dropna().unique())

table = []

for year in years:

    year_apartments = kyiv_three_rooms[
        kyiv_three_rooms["year"] == year
    ]

    count = len(year_apartments)

    average_days = year_apartments["days_exposition"].mean()

    table.append([year, count, round(average_days, 2)])

result = pd.DataFrame(
    table,
    columns=["Year", "Number of apartments", "Average days exposition"]
)

result
```

С 2021 по 2024 год количество объявлений о продаже трёхкомнатных квартир в Киеве значительно увеличилось.
При этом среднее время размещения объявлений сократилось.
Это может свидетельствовать о повышении активности рынка недвижимости и более быстром исчезновении объявлений с площадки. Однако нельзя утверждать, что все эти квартиры были проданы, поскольку в датасете нет информации о завершённых сделках.

```python
apartments["year"] = apartments["first_day_exposition"].str[:4]

kyiv = apartments[apartments["locality_name"] == "Kyiv"]

years = sorted(kyiv["year"].dropna().unique())

rooms = sorted(kyiv["rooms"].dropna().unique())

table = []

for year in years:

    year_apartments = kyiv[kyiv["year"] == year]

    row = []

    for room in rooms:

        count = len(year_apartments[year_apartments["rooms"] == room])

        row.append(count)

    table.append(row)

rooms_year = pd.DataFrame(table, index=years, columns=rooms)

rooms_year
```

В Киеве наибольшей популярностью пользуются одно-, двух- и трёхкомнатные квартиры. В 2024 году по сравнению с 2023 годом увеличилась доля объявлений об однокомнатных и двухкомнатных квартирах, а доля трёхкомнатных немного снизилась.
Можно предположить, это связано с тем, что небольшие квартиры могут быть более доступными по цене и требуют меньших расходов на содержание. Также покупатели могут выбирать компактное жильё из-за изменения своих потребностей и финансовых возможностей.
Таким образом, в последнее время заметен рост интереса к одно- и двухкомнатным квартирам.
