```python
import pandas as pd

orders1 = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/orders_part_1.csv")
orders2 = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/orders_part_2.csv")
customers = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/customers.csv")

orders = pd.concat([orders1, orders2], ignore_index=True)

orders = pd.DataFrame(orders)
customers = pd.DataFrame(customers)

data = orders.join(customers.set_index("CustomerID"), on="CustomerID", how="inner")

data
```

1

```python
result = data.groupby(["Location", "Category"]).size().unstack(fill_value=0)

result
```

2

```python
gaming = data[data["Description"] == "Gaming Keyboard"]

result = gaming.groupby("Location").agg(
    Number_of_orders=("InvoiceNo", "count"),
    Number_of_keyboards=("Quantity", "sum")
)

result = result.sort_values(by="Number_of_keyboards", ascending=False)

result.head(1)
```

3

```python
rich_customers = data[data["Annual_Income"] > 2000]

result = rich_customers.groupby("Category").size()

result = result.sort_values(ascending=False)

result
```
