```python
n = int(input("Enter a number between 5 and 15: "))

if n > 4 and n < 16:
  fact = 1

  for i in range(n, 0, -1):
    fact = fact * i

  print("The factorial is equal to: ", fact)

else:
  print("Invalid. The number must be between 5 and 15.")
```

```python
height = int(input("Enter the hight: "))
width = int(input("Enter the width: "))

i = 1

while i <= height:
  if i == 1 or i == height:
    print("*" * width)

  else:
    print("*" + " " * (width - 2) + "*")

  i = i + 1
```

```python
width = int(input("Enter an odd number: "))

if width % 2 == 1:

  for stars in range(width, 0, -2):
    stars_space = (width - stars) // 2
    print(" " * stars_space + "*" * stars)

  for stars in range(3, width + 1, 2):
    stars_space = (width - stars) // 2
    print(" " * stars_space + "*" * stars)

else:
  print("An even number")
```
