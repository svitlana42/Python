```python
import random

salaries = []

for salary in range(12):
  salaries.append(random.randint(7500, 15000))

print("Monthly salary: ", salaries)

s = 0

for money in salaries:
  s += money

result = s / len(salaries)

print("Average salary: ", result)
```

Зеркальный переворот

```python
number = [7, 2, 9, 4]

for a in range(len(number) // 2):
  coup = number[a]
  number[a] = number[len(number) - 1 - a]
  number[len(number) - 1 - a] = coup

print(number)
```
