```python
text = input("Enter a string: ")

words = text.split()

long = words[0]

for word in words:
  if len(word) > len(long):
    long = word

print("The longest word: ", long)
```

```python
text = input("Enter a string containing letters and numbers: ")

text1 = ""

for letter in text:
  if letter.isdigit():
    text1 += "*"
  else:
    text1 += letter

print(text1)

```

```python
a = input("Enter a string: ")

b = len(a)
result = a

for c in range(1, b + 1):
  if b % c == 0:
    word = a[:c]

    if word * (b // c) == a:
      result = word
      break

print(f"Word: {result}")

```
