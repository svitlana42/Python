```python
text = input("Enter a first word: ")
text1 = input("Enter a second word: ")

letters = set(text) & set(text1)

if len(letters) == 0:
  print("There are no identical letters")

else:
  print("Identical letters: ")
  for letter in letters:
    print(letter)
```

```python
num = [i for i in range(1, 101) if i % 3 == 0]
num1 = [i for i in range(1, 101) if i % 5 == 0]

result = set(num) & set(num1)

print("Multiples of three: ", num)

print("Multiples of five: ", num1)

print("Identical: ", sorted(result))
```
