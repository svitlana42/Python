```python
day_week = {
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday",
    7: "Sunday"
}

num = int(input("Enter the day of the week: "))

if num in day_week:
  print(day_week[num])
else:
  print("There is no such day of the week")
```

```python
text = input("Enter the text: ")

letters = {}

for letter in text:
  if letter != " ":
    if letter in letters:
      letters[letter] += 1
    else:
      letters[letter] = 1

for key, value in letters.items():
  print(key, "-", value)
```

```python
teams = [
    {"name": "Olga", "department": "hr"},
    {"name": "Inna", "department": "test"},
    {"name": "Alex", "department": "developer"},
    {"name": "Anna", "department": "developer"},
    {"name": "Ira", "department": "hr"}
]

sections = {}

for team in teams:
  s = team["department"]
  name = team["name"]

  if s not in sections:
    sections[s] = []

  sections[s].append(name)

print(sections)

```
