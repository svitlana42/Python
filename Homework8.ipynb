````python
import csv
import math
from google.colab import drive

drive.mount('/content/drive')

with open("/content/drive/MyDrive/Colab Notebooks/Lection_8_hw.csv", "r", encoding="utf-8") as file:

  reader = csv.DictReader(file)

  phones = {}

  for phone in reader:
    price = phone["price_range"]

    if price not in phones:
      phones[price] = []

    phones[price].append(phone)

new_phones = []

for price in phones:
  phone_list = phones[price]

  for phone in phone_list:

    if phone["sc_w"] != "" and phone["sc_h"] != "":
      width = float(phone['sc_w'])
      height = float(phone['sc_h'])

      # вычисляю диагональ и округляю до сотых
      phone["diagonal"] = round(math.sqrt(width ** 2 + height ** 2), 2)

      # удаляю ширину и высоту
      del phone ['sc_w']
      del phone ['sc_h']

      new_phones.append(phone)

# записываю новый csv
with open("/content/drive/MyDrive/Colab Notebooks/phones.csv", "w", newline="", encoding="utf-8") as file:

  writer = csv.writer(file)

  writer.writerow(new_phones[0].keys())

  for phone in new_phones:
    writer.writerow(phone.values())

for phone in new_phones:
  print(phone)



````
