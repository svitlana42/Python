```python
def group(employees):

  departments = {}

  for employee in employees:
    department = employee["department"]
    name = employee["name"]

    if department not in departments:
      departments[department] = []

    departments[department].append(name)

  return departments

employees = [
    {"name": "Olga", "department": "hr"},
    {"name": "Inna", "department": "test"},
    {"name": "Alex", "department": "developer"},
    {"name": "Anna", "department": "developer"},
    {"name": "Ira", "department": "hr"}
]

print(group(employees))
```

```python
def above_threshold(numbers, number):

  count = 0

  for item in numbers:
    if item > number:
      count += 1

  return count

result = above_threshold([5, 9, 14, 23], 7)

print(result)
```

```python
def count_keyword_occurrences(strings, word):

  count = 0

  for string in strings:
    words = string.split()

    for value in words:
      if value == word:
        count += 1

  return count

result = count_keyword_occurrences([
    "Hello world",
    "The weather is nice today",
    "The weather will be bad tomorrow"
], "weather")

print(result)
```

```python
def new_number(sequence):

  numbers = sequence.split()

  for i in range(len(numbers)):
    numbers[i] = int(numbers[i])

  # нахожу разницу между числами
  differences = []

  for i in range(1, len(numbers)):
    differences.append(numbers[i] - numbers[i - 1])

  # если разница одинаковая
  if differences.count(differences[0]) == len(differences):
    return numbers[-1] + differences[-1]

  # проверка, увеличивается ли на одно и то же число
  if len(differences) >= 3:

    difference = differences[1] - differences[0]

    if difference == differences[2] - differences[1]:
      next = differences[-1] + difference
      return numbers[-1] + next

  # проверяю умножение
  if numbers[0] != 0:
    mult = numbers[1] / numbers[0]

    if numbers[-1] / numbers[-2] == mult:
      return int(numbers[-1] * mult)

 # проверяю квадраты
  square = True

  for i in range(len(numbers)):
    if numbers[i] != (i + 1) ** 2:
      square = False
      break

  if square:
    next = len(numbers) + 1
    return next ** 2

# проверяю кубы
  cube = True

  for i in range(len(numbers)):
    if numbers[i] != (i + 1) ** 3:
      cube = False
      break

  if cube:
    next = len(numbers) + 1
    return next ** 3

sequence = input("Введите последовательность чисел через пробел: ")

print("Следующее число: ", new_number(sequence))
```

```python
def palindrome(number):
  # перевожу число в строку и сравниваю ее с ней же перевернутой
  return str(number) == str(number)[::-1]

max_palindrome = 0
num1 = 0
num2 = 0

# перебираю 3-х значные числа
for i in range(100, 1000):
  for j in range(i, 1000):
    result = i * j

    # если палиндром больше максимума
    if palindrome(result):
      if result > max_palindrome:
        max_palindrome = result
        num1 = i
        num2 = j

print(f"Самый большой палиндром: {max_palindrome}")
print(f"Палиндром является произведением чисел: {num1} и {num2}")

```
